```python
import subprocess

c_code = """
#include <stdio.h>

#define INF 99999
#define V 4

void floydWarshall(int graph[V][V])
{
    int dist[V][V];

    // Initialize distance matrix
    for (int i = 0; i < V; i++)
    {
        for (int j = 0; j < V; j++)
        {
            dist[i][j] = graph[i][j];
        }
    }

    // Floyd-Warshall Algorithm
    for (int k = 0; k < V; k++)
    {
        for (int i = 0; i < V; i++)
        {
            for (int j = 0; j < V; j++)
            {
                if (dist[i][k] + dist[k][j] < dist[i][j])
                {
                    dist[i][j] = dist[i][k] + dist[k][j];
                }
            }
        }
    }

    // Print shortest distance matrix
    printf("Shortest Distance Matrix:\\n");

    for (int i = 0; i < V; i++)
    {
        for (int j = 0; j < V; j++)
        {
            printf("%7d", dist[i][j]);
        }
        printf("\\n");
    }
}

int main()
{
    int graph[V][V] = {
        {0, 5, INF, 10},
        {INF, 0, 3, INF},
        {INF, INF, 0, 1},
        {INF, INF, INF, 0}
    };

    floydWarshall(graph);

    return 0;
}
"""

# Write the C code into a file
with open("floyd.c", "w") as f:
    f.write(c_code)

# Compile the C file
compilation = subprocess.run(["gcc", "floyd.c", "-o", "floyd"], capture_output=True, text=True)

if compilation.returncode == 0:
    # Run the compiled binary
    execution = subprocess.run(["./floyd"], capture_output=True, text=True)
    print(execution.stdout)
else:
    print("Compilation failed:")
    print(compilation.stderr)
```
